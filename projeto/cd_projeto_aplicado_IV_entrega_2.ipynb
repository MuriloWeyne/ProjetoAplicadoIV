# **Referencial Teórico**

Uma série temporal é definida como um conjunto de observações coletadas sequencialmente ao longo do tempo em intervalos regulares (SILVA; PERES; BOSCARIOLI, 2016). A análise de séries temporais tem por objetivo identificar padrões e extrair características significativas dos dados, possibilitando a realização de previsões e o suporte à tomada de decisão em diferentes domínios, como finanças, saúde e comércio.

Os principais componentes de uma série temporal são: (i) **tendência**, que representa a direção de longo prazo dos dados (crescente, decrescente ou estável); (ii) **sazonalidade**, que corresponde a padrões de variação regular que se repetem em intervalos fixos de tempo — como o aumento das vendas do varejo no período natalino; (iii) **ciclos**, variações repetitivas cujo período não é fixo; e (iv) **ruído** (erro aleatório), variações irregulares e inexplicáveis nos dados (SILVA; PERES; BOSCARIOLI, 2016). Uma série é classificada como **estacionária** quando sua média e variância permanecem constantes ao longo do tempo. Séries sazonais e cíclicas, em geral, não possuem caráter estacionário, sendo necessário aplicar transformações — como diferenciação — para torná-las estacionárias antes da modelagem.

Entre os modelos estatísticos clássicos para previsão de séries temporais, destaca-se o **ARIMA** (*Autoregressive Integrated Moving Average*), que combina componentes autorregressivos (AR), de integração (I) e de médias móveis (MA). Uma extensão direta, o **SARIMA** (*Seasonal ARIMA*), incorpora explicitamente a componente sazonal ao modelo, sendo especialmente adequado para séries com padrões de sazonalidade regulares, como o índice de volume de vendas do varejo (BOX et al., 2015). As principais vantagens do SARIMA são a interpretabilidade dos parâmetros e a capacidade de modelar tanto a dependência temporal quanto a sazonalidade. Suas limitações incluem a pressuposição de linearidade e a sensibilidade à escolha manual dos hiperparâmetros (p, d, q, P, D, Q, s).

Para capturar padrões não lineares e dependências de longo prazo em sequências temporais, modelos de aprendizado profundo têm sido amplamente adotados. As **Redes Neurais Recorrentes** (*Recurrent Neural Networks* — RNN) processam sequências por meio de uma arquitetura com retroalimentação, na qual a saída de uma etapa é utilizada como entrada da etapa seguinte, conferindo ao modelo uma capacidade de "memória" sobre observações anteriores (IBM, s.d.). A variante **LSTM** (*Long Short-Term Memory*), proposta por Hochreiter e Schmidhuber (1997), supera as RNNs simples ao mitigar o problema do desaparecimento do gradiente por meio de portas de controle (entrada, esquecimento e saída), tornando-a mais eficaz na captura de dependências de longo prazo em séries temporais. As vantagens das LSTMs incluem a capacidade de modelar não linearidades e a ausência de requisito de estacionaridade. Como limitações, exigem maior volume de dados para treinamento, maior custo computacional e apresentam menor interpretabilidade em comparação aos modelos estatísticos tradicionais.

No contexto do varejo brasileiro, Bagatini e Laimer (2019) analisaram o desempenho do varejo online no Brasil entre 2003 e 2018, documentando crescimento estrutural consistente no período e evidenciando a relevância do monitoramento temporal das vendas. O presente projeto expande essa perspectiva ao empregar tanto abordagens estatísticas clássicas (SARIMA) quanto de aprendizado profundo (LSTM) para a previsão do volume de vendas do varejo paulista, com dados oficiais da PMC/IBGE abrangendo o período de 2000 a 2025.

# **Pipeline da Solução**

O pipeline proposto para a análise e previsão do volume de vendas do comércio varejista de São Paulo compreende cinco etapas sequenciais, conforme ilustrado abaixo:

```
Coleta dos dados → Pré-processamento → EDA & Visualização → Modelagem → Avaliação e Validação
```

**1. Coleta dos dados**

Os dados são obtidos diretamente da API pública do IBGE, referentes à tabela 8880 da Pesquisa Mensal de Comércio. A coleta é automatizada via requisições HTTP ao endpoint da API SIDRA, recuperando o índice de volume de vendas no comércio varejista do estado de São Paulo no período de janeiro de 2000 a dezembro de 2025 (312 observações mensais).

**2. Pré-processamento**

Esta etapa envolve: conversão das datas para o formato `datetime` com frequência mensal; verificação e tratamento de eventuais valores faltantes (por interpolação linear ou preenchimento por valores vizinhos); conversão dos valores do índice para tipo numérico; e organização da série com o índice temporal como referência. Por se tratar de um índice de volume deflacionado, não é necessário aplicar deflação adicional.

**3. Análise Exploratória dos Dados (EDA) e Visualização**

São produzidas as seguintes análises: (i) gráfico de linha da série histórica completa; (ii) decomposição da série em tendência, sazonalidade e resíduos (decomposição aditiva e multiplicativa); (iii) boxplot por mês do ano para visualização do padrão sazonal; (iv) mapa de calor mensal ao longo dos anos; e (v) análise de estacionaridade por meio do teste Augmented Dickey-Fuller (ADF), com aplicação de diferenciação caso necessário.

**4. Modelagem**

Serão desenvolvidos e comparados dois modelos preditivos:

- **SARIMA** (*Seasonal Autoregressive Integrated Moving Average*): modelo estatístico clássico que incorpora as componentes autorregressiva, de integração e de médias móveis, com extensão sazonal. Servirá como modelo de referência (*baseline*). Os hiperparâmetros (p, d, q, P, D, Q, s) serão selecionados com base nas funções de autocorrelação (ACF) e autocorrelação parcial (PACF), complementadas por busca em grid com critério AIC.
- **LSTM** (*Long Short-Term Memory*): rede neural recorrente com portas de memória, capaz de capturar padrões não lineares e dependências de longo prazo na série temporal. A série será normalizada e estruturada em janelas deslizantes para o treinamento supervisionado.

**5. Avaliação e Validação**

Os modelos serão avaliados por meio de divisão temporal dos dados em conjuntos de treino (jan/2000 a dez/2022) e teste (jan/2023 a dez/2025), utilizando as métricas:

| Métrica | Descrição |
|---|---|
| MAE | *Mean Absolute Error* — erro absoluto médio |
| RMSE | *Root Mean Squared Error* — raiz do erro quadrático médio |
| MAPE | *Mean Absolute Percentage Error* — erro percentual absoluto médio |

Os resultados serão apresentados em tabela comparativa entre os modelos SARIMA e LSTM.
